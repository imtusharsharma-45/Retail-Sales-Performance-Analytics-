# Visualization

```python
#Understand:

#Most orders small hain ya large?
```

```python
#Histogram
plt.figure(figsize=(10,5))
plt.hist(df['Sales'], bins=30)
plt.title("Sales Distribution")
plt.xlabel("sales")
plt.ylabel("Number of orders")

plt.show() 
```

```python
# Observation:
# Most orders have low sales values, while a few orders generate extremely high sales.

# Insight:
# The sales distribution is right-skewed, indicating dependence on a small number of high-value orders.

# Recommendation:
# The business should identify high-value customers and products to maximize revenue opportunities.
```

# Profit Distribution Analysis

```python

# profit healthy hai ya losses zyada hain.
```

```python
#Profit Histogram 
plt.figure(figsize=(10,5))

plt.hist(df['Profit'], bins=30)

plt.title("Profit Distribution")
plt.xlabel("Profit")
plt.ylabel("Number of Orders")

plt.show()
```

```python
#Observation:
#Most orders generate small profits, while some orders result in heavy losses.

#Insight:
#Negative profit orders indicate that certain products or discount strategies may be hurting profitability.

#Recommendation:
#The company should analyze loss-making products and optimize discount policies.
```

# Top Performing Categories

# Category-wise Profit Analysis

```python
#Category-wise Profit
category_profit = df.groupby('Category')['Profit'].sum()

category_profit
```

```python
#Profit Bar Chart
category_profit.plot(kind='bar', figsize=(8,5))

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.show()
```

```python
## Observation
#Technology category generates the highest profit among all categories.

## Insight
#Technology products are both strong revenue and profit drivers for the business.

## Recommendation
#The company should focus more on high-performing technology products and customers.
```

# Sub-Category Sales Analysis

```python
#Sub-Category Sales
subcat_sales = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)

subcat_sales
```

```python
#Top 10 Sub-Categories Chart
top10_subcat = subcat_sales.head(10)

top10_subcat.plot(kind='bar', figsize=(10,5))

plt.title("Top 10 Sub-Categories by Sales")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")

plt.show()
```

# Top Profitable Sub-Categories

```python
#Sub-Category Profit
subcat_profit = df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False)

subcat_profit
```

```python
#Top 10 Profit Chart
top10_profit = subcat_profit.head(10)

top10_profit.plot(kind='bar', figsize=(10,5))

plt.title("Top 10 Sub-Categories by Profit")
plt.xlabel("Sub-Category")
plt.ylabel("Profit")

plt.show()
```

```python
### Observation
#Phones generate the highest sales, while Copiers generate the highest profit.

## Insight
#High-sales products are not always the most profitable. Copiers appear to have stronger profit margins compared to Phones.

## Recommendation
#The company should balance revenue growth with profitability by promoting high-margin products like Copiers.
```

# Loss-Making Sub-Categories 

```python
#Bottom Profit Sub-Categories
bottom_profit = df.groupby('Sub-Category')['Profit'].sum().sort_values()

bottom_profit
```

```python
#Bottom 10 Profit Chart
bottom10 = bottom_profit.head(10)

bottom10.plot(kind='bar', figsize=(10,5))

plt.title("Bottom 10 Sub-Categories by Profit")
plt.xlabel("Sub-Category")
plt.ylabel("Profit")

plt.show()
```

```python
## Observation
#Tables is the most loss-making sub-category.

## Insight
#Despite generating sales, Tables may suffer from low margins or excessive discounting, resulting in significant losses.

## Recommendation
#The business should review pricing, discounts, and operational costs for the Tables sub-category.
```

# Monthly Sales Trend Analysis 

```python
# Month-Year Column 
df['Month_Year'] = df['Order Date'].dt.to_period('M')
```

```python
#Monthly Sales Calculate
monthly_sales = df.groupby('Month_Year')['Sales'].sum()
```

```python
#Line Chart
plt.Figure(figsize=(12,5))
monthly_sales.plot()
plt.title("Monthly Sales Trend")
plt.xlabel("Month-Year")
plt.ylabel("Sales")

plt.show()
```

```python
## Observation
#Sales fluctuate over time with noticeable peaks during certain periods.

## Insight
#The business experiences seasonal sales trends and overall growth over time.

## Recommendation
#The company should prepare inventory and marketing strategies for high-demand periods.
```

# Top Customers Analysis

```python
#Top Customers
top_customers = df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False)

top_customers.head(10)
```

```python
#Top 10 Customers Chart
top10_customers = top_customers.head(10)

top10_customers.plot(kind='bar', figsize=(12,5))

plt.title("Top 10 Customers by Sales")
plt.xlabel("Customer Name")
plt.ylabel("Sales")

plt.show()
```

```python
## Observation
#A small group of customers contributes significantly to total sales revenue.

## Insight
#The business is highly dependent on high-value customers such as Sean Miller.

## Recommendation
#The company should focus on customer retention strategies and personalized engagement for top customers.
```

# Final Business Insights

## Key Findings

- Technology is the highest-performing category in both sales and profit.
- Phones generate the highest sales among all sub-categories.
- Copiers are the most profitable sub-category.
- Tables generate the highest losses for the business.
- The West region contributes the highest sales and profit.
- The Central region shows comparatively weak profitability.
- Sales show an overall upward trend over time.
- Discounts negatively impact profitability.
- A small number of customers contribute significantly to total revenue.

---

## Business Recommendations

- Focus more on high-performing technology products.
- Promote high-margin products like Copiers.
- Review pricing and discount strategies for loss-making products like Tables.
- Optimize business operations in low-profit regions.
- Improve customer retention strategies for high-value customers.
- Carefully manage discount policies to protect profit margins.
